```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([0, 1, 1, 0])
```

```python
class Perceptron:
    def __init__(self, learning_rate=0.1, epochs=2):
        self.lr = learning_rate
        self.epochs = epochs
        self.weights = None
        self.bias = None

    def _activation(self, z):
        return np.where(z >= 0, 1, 0)

    def fit(self, X, y):
        self.weights = np.zeros(X.shape[1])
        self.bias = 0.0

        print("Starting training...")
        for epoch in range(self.epochs):
            errors_in_epoch = 0
            for idx, x_i in enumerate(X):
                non_linear_output = np.dot(x_i, self.weights) + self.bias
                prediction = self._activation(non_linear_output)

                error = y[idx] - prediction

                if error != 0:
                    self.weights += self.lr * error * x_i
                    self.bias += self.lr * error
                    errors_in_epoch += 1

            print(f"Epoch {epoch+1}/{self.epochs} - Classification Errors: {errors_in_epoch}")

            if errors_in_epoch == 0:
                print(f"--> Converged successfully at Epoch {epoch+1}!")
                break

    def predict(self, X):
        non_linear_output = np.dot(X, self.weights) + self.bias
        return self._activation(non_linear_output)
```

```python
clf = Perceptron(learning_rate=0.1, epochs=10)
clf.fit(X, y)

predictions = clf.predict(X)

print("\nActual Output:", y)
print("Predicted Output:", predictions)
```
