```python
import numpy as np
```

```python
class NeuralNetwork:
    # Initialise random weights and zero biases
    def __init__(self, i_size, h_size, o_size):
        self.w1, self.b1 = np.random.randn(i_size, h_size), np.zeros((1, h_size))
        self.w2, self.b2 = np.random.randn(h_size, o_size), np.zeros((1, o_size))

    # Forward pass through both layers with sigmoid activation
    def forward(self, x):
        sig = lambda x: 1 / (1 + np.exp(-x))
        return sig(sig(x @ self.w1 + self.b1) @ self.w2 + self.b2)
```

```python
if __name__ == "__main__":
    # Create network
    nn = NeuralNetwork(i_size=3, h_size=5, o_size=2)

    # Sample input
    X_input = np.array([
        [0.1, 0.5, 0.9],
        [0.2, 0.8, 0.6]
    ])

    # Run forward pass
    output = nn.forward(X_input)

    # Print results
    print("Output shape:", output.shape)
    print("Network predictions:\n", output)
```
