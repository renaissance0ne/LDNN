```python
import numpy as np
```

```python
# Hyperbolic tangent sigmoid activation function
def tansig(Z):
  return (2 / (1 + np.exp(-2 * Z))) - 1
```

```python
# Inputs, weights and bias
X=np.array([0.5,-1.2,3.3])
W = np.array([0.8, -0.5, 1.0])
b = 0.2

# Weighted sum and activation
z=np.dot(W,X)+b

output_numpy_tanh = tansig(z)

# Print results
print("Weighted sum(Z):",z)
print("Output with numpy tanh:", output_numpy_tanh)
```
