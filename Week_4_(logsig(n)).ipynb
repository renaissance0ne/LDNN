```python
import numpy as np
import math
```

```python
# Logistic sigmoid activation function
def logsig(z):
  return 1 / (1 + math.exp(-z))
```

```python
# Inputs, weights and bias
X=np.array([0.5,-1.2,3.3])
W = np.array([0.8, -0.5, 1.0])
b = 0.2

# Weighted sum and activation
z=np.dot(W,X)+b
op_log=logsig(z)

# Print results
print("Weighted sum(z):",z)
print("Output log(nōn-linear):", op_log)
```
